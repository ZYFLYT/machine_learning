# 简单图形化测试

这是一个用于测试的简单 notebook,演示使用 matplotlib 绘制几种常见图表。

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. 折线图

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.xlabel('x')
plt.title('正弦与余弦函数')
plt.legend()
plt.grid(True)
plt.ylabel('y')
plt.show()
```

## 2. 散点图

```python
rng = np.random.default_rng(42)
n = 100
x = rng.normal(0, 1, n)
y = rng.normal(0, 1, n)

plt.figure(figsize=(6, 4))
plt.scatter(x, y, c=x + y, cmap='viridis', alpha=0.7)
plt.colorbar(label='x + y')
plt.xlabel('x')
plt.ylabel('y')
plt.title('随机散点图')
plt.show()
```

## 3. 柱状图

```python
categories = ['A', 'B', 'C', 'D', 'E']
values = [12, 25, 8, 30, 17]

plt.figure(figsize=(6, 4))
plt.bar(categories, values, color=['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8172B3'])
plt.xlabel('类别')
plt.ylabel('数值')
plt.title('柱状图示例')
plt.show()
```
